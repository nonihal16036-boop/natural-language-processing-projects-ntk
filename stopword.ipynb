```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
# check whether a word is a stopword
words = ['the','is','python','and','beautifull']
for word in words:
  doc = nlp(word)
for token in doc:
    print(token.text,"->",token.is_stop)
```

```python
# extract only stopwords
import spacy
nlp = spacy.load("en_core_web_sm")
words = ['the','is','python','and','beautifull']
for word in words:
  doc = nlp(word)
for token in doc:
    if token.is_stop:
      stopwords.append(token.text)
      print(stopwords)
```

```python
# extract non stopwords
import spacy
nlp = spacy.load("en_core_web_sm")
text = "python is an easy and powerful language."
doc = nlp(text)
words = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.text)
print(words)
```

```python
import nltk
nltk.download('punkt_tab')

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

stop_words = set(stopwords.words('english'))

text = "Python is an easy and powerful language."

tokens = word_tokenize(text)

result = []

for word in tokens:
    if word.lower() not in stop_words and word not in string.punctuation:
        result.append(word)

print(result)
```
