```python
import spacy
nlp = spacy.load("en_core_web_sm")
text = "apple is headquartered in california."
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Apple is headquartered in California."
doc = nlp(text)

for ent in doc.ents:
    if ent.label_ == "PERSON":
        print(ent.text)
```
